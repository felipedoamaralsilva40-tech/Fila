```python
class AFD:
    def __init__(self, estados, alfabeto, inicial, finais, transicoes, nome="AFD"):
        self.estados = set(estados)
        self.alfabeto = set(alfabeto)
        self.inicial = inicial
        self.finais = set(finais)
        self.transicoes = dict(transicoes)
        self.nome = nome

    def executar(self, palavra, mostrar_passos=False):
        estado = self.inicial
        passos = [estado]

        for caractere in palavra:
            destino = self.transicoes.get((estado, caractere))
            if destino is None:
                if mostrar_passos:
                    return False, passos, f"Não existe transição de {estado} lendo {caractere!r}."
                return False
            estado = destino
            passos.append(estado)

        aceita = estado in self.finais
        if mostrar_passos:
            motivo = "Entrada aceita." if aceita else f"A entrada terminou em {estado}, que não é estado final."
            return aceita, passos, motivo
        return aceita

    def tabela(self):
        simbolos = sorted(self.alfabeto)
        linhas = [f"AFD: {self.nome}", "Estado | " + " | ".join(simbolos)]
        for estado in sorted(self.estados):
            inicio = "-> " if estado == self.inicial else "   "
            final = " *" if estado in self.finais else ""
            destinos = [str(self.transicoes.get((estado, s), "-")) for s in simbolos]
            linhas.append(inicio + estado + final + " | " + " | ".join(destinos))
        return "\n".join(linhas)
```

```python
estados_senha = ["q0", "q1", "q2", "q3", "q4", "erro"]
alfabeto_senha = set("PN0123456789?")
transicoes_senha = {}

transicoes_senha[("q0", "P")] = "q1"
transicoes_senha[("q0", "N")] = "q1"
for c in alfabeto_senha - {"P", "N"}:
    transicoes_senha[("q0", c)] = "erro"

for c in "0123456789":
    transicoes_senha[("q1", c)] = "q2"
    transicoes_senha[("q2", c)] = "q3"
    transicoes_senha[("q3", c)] = "q4"

for estado in ["q1", "q2", "q3"]:
    for c in alfabeto_senha:
        transicoes_senha.setdefault((estado, c), "erro")
for c in alfabeto_senha:
    transicoes_senha[("q4", c)] = "erro"
    transicoes_senha[("erro", c)] = "erro"

afd_senha = AFD(
    estados_senha, alfabeto_senha, "q0", {"q4"}, transicoes_senha,
    "Senha P000 ou N000"
)

def validar_senha(texto, mostrar_passos=False):
    convertido = "".join(c if c in alfabeto_senha else "?" for c in texto)
    return afd_senha.executar(convertido, mostrar_passos)

print(afd_senha.tabela())
print("\nP042:", validar_senha("P042"))
print("N117:", validar_senha("N117"))
print("X042:", validar_senha("X042"))
print("P42:", validar_senha("P42"))
```

```python
estados_guiche = ["g0", "g1", "g2", "erro"]
alfabeto_guiche = set("G0123456789?")
transicoes_guiche = {("g0", "G"): "g1"}

for c in alfabeto_guiche - {"G"}:
    transicoes_guiche[("g0", c)] = "erro"
for c in "123456789":
    transicoes_guiche[("g1", c)] = "g2"
for c in alfabeto_guiche - set("123456789"):
    transicoes_guiche[("g1", c)] = "erro"
for c in alfabeto_guiche:
    transicoes_guiche[("g2", c)] = "erro"
    transicoes_guiche[("erro", c)] = "erro"

afd_guiche = AFD(
    estados_guiche, alfabeto_guiche, "g0", {"g2"}, transicoes_guiche,
    "Guichê G1 a G9"
)

def validar_guiche(texto, mostrar_passos=False):
    convertido = "".join(c if c in alfabeto_guiche else "?" for c in texto)
    return afd_guiche.executar(convertido, mostrar_passos)

print(afd_guiche.tabela())
print("\nG1:", validar_guiche("G1"))
print("G9:", validar_guiche("G9"))
print("G0:", validar_guiche("G0"))
print("G10:", validar_guiche("G10"))
```

```python
def mostrar_rastro(nome, funcao, entrada):
    aceita, passos, motivo = funcao(entrada, mostrar_passos=True)
    print(f"{nome}: {entrada!r}")
    print("Resultado:", "ACEITA" if aceita else "REJEITADA")
    print("Caminho:", " -> ".join(passos))
    print("Motivo:", motivo, "\n")

mostrar_rastro("Senha", validar_senha, "P042")
mostrar_rastro("Senha", validar_senha, "P42")
mostrar_rastro("Guichê", validar_guiche, "G3")
mostrar_rastro("Guichê", validar_guiche, "G10")
```

```python
regex_senha = re.compile(r"[PN][0-9]{3}")
regex_guiche = re.compile(r"G[1-9]")

testes_senha = [
    ("P042", True), ("N117", True), ("P000", True), ("N999", True), ("P123", True),
    ("", False), ("P42", False), ("X042", False), ("P1234", False), ("N1A7", False)
]
testes_guiche = [
    ("G1", True), ("G3", True), ("G5", True), ("G8", True), ("G9", True),
    ("", False), ("G0", False), ("G10", False), ("g3", False), ("H2", False)
]

def comparar(nome, testes, funcao_afd, regex):
    divergencias = []
    print(f"--- {nome} ---")
    for entrada, esperado in testes:
        resultado_afd = funcao_afd(entrada)
        resultado_regex = regex.fullmatch(entrada) is not None
        print(f"{entrada!r:8} AFD={resultado_afd!s:5} RegEx={resultado_regex!s:5}")
        if resultado_afd != resultado_regex or resultado_afd != esperado:
            divergencias.append(entrada)
    print("Total de divergências:", len(divergencias), divergencias, "\n")

comparar("Senhas", testes_senha, validar_senha, regex_senha)
comparar("Guichês", testes_guiche, validar_guiche, regex_guiche)
```

```python
class Atendimento:
    def __init__(self, senha):
        if not validar_senha(senha):
            raise ValueError("Senha inválida. Use P042 ou N117.")
        self.senha = senha
        self.estado = "EMITIDA"
        self.transicoes = {
            ("EMITIDA", "chamar"): "CHAMADA",
            ("CHAMADA", "iniciar_atendimento"): "EM_ATENDIMENTO",
            ("CHAMADA", "nao_compareceu"): "AUSENTE",
            ("EM_ATENDIMENTO", "finalizar"): "FINALIZADA",
        }

    def eventos_permitidos(self):
        return [evento for (estado, evento) in self.transicoes if estado == self.estado]

    def aplicar(self, evento):
        chave = (self.estado, evento)
        if chave not in self.transicoes:
            permitidos = self.eventos_permitidos()
            print(f"Evento inválido {evento!r} para {self.senha} no estado {self.estado}.")
            print("Eventos permitidos:", permitidos if permitidos else "nenhum")
            return False
        anterior = self.estado
        self.estado = self.transicoes[chave]
        print(f"{self.senha}: {anterior} --{evento}--> {self.estado}")
        return True

def executar_fluxo(senha, eventos):
    atendimento = Atendimento(senha)
    print("Início:", atendimento.senha, atendimento.estado)
    for evento in eventos:
        if not atendimento.aplicar(evento):
            break
    print("Estado final:", atendimento.estado, "\n")

print("Fluxo válido 1 — atendimento concluído:")
executar_fluxo("P042", ["chamar", "iniciar_atendimento", "finalizar"])

print("Fluxo válido 2 — senha ausente:")
executar_fluxo("N117", ["chamar", "nao_compareceu"])

print("Fluxo inválido 1 — tentar finalizar sem chamar:")
executar_fluxo("P042", ["finalizar"])

print("Fluxo inválido 2 — chamar duas vezes:")
executar_fluxo("N117", ["chamar", "chamar"])
```

```python
PALAVRAS_RESERVADAS = {
    "EMITIR", "PREFERENCIAL", "NORMAL", "CHAMAR", "GUICHE",
    "INICIAR", "FINALIZAR", "NAO_COMPARECEU"
}

def scanner_fila(codigo):
    tokens, erros = [], []
    i, linha, coluna = 0, 1, 1
    n = len(codigo)

    def avancar(texto):
        nonlocal linha, coluna
        partes = texto.split("\n")
        if len(partes) > 1:
            linha += len(partes) - 1
            coluna = len(partes[-1]) + 1
        else:
            coluna += len(texto)

    while i < n:
        c = codigo[i]
        if c.isspace():
            avancar(c)
            i += 1
            continue
        if c == "#":
            inicio = i
            while i < n and codigo[i] != "\n":
                i += 1
            avancar(codigo[inicio:i])
            continue

        inicio = i
        linha_inicio, coluna_inicio = linha, coluna

        if c.isalpha() or c == "_":
            i += 1
            while i < n and (codigo[i].isalnum() or codigo[i] == "_"):
                i += 1
            lexema = codigo[inicio:i]
            maiusculo = lexema.upper()
            if maiusculo in PALAVRAS_RESERVADAS:
                tipo, valor = "PALAVRA_RESERVADA", maiusculo
            elif re.fullmatch(r"[PN][0-9]{3}", maiusculo):
                tipo, valor = "SENHA", maiusculo
            elif re.fullmatch(r"G[1-9]", maiusculo):
                tipo, valor = "GUICHE", maiusculo
            else:
                tipo, valor = "IDENTIFICADOR", lexema
            tokens.append({
                "tipo": tipo, "lexema": valor,
                "linha": linha_inicio, "coluna": coluna_inicio
            })
            avancar(codigo[inicio:i])
            continue

        if c.isdigit():
            i += 1
            while i < n and codigo[i].isdigit():
                i += 1
            lexema = codigo[inicio:i]
            tokens.append({
                "tipo": "NUMERO", "lexema": lexema,
                "linha": linha_inicio, "coluna": coluna_inicio
            })
            avancar(lexema)
            continue

        simbolos = {
            ";": "PONTO_E_VIRGULA", ",": "VIRGULA",
            ":": "DOIS_PONTOS", "-": "HIFEN"
        }
        if c in simbolos:
            tokens.append({
                "tipo": simbolos[c], "lexema": c,
                "linha": linha_inicio, "coluna": coluna_inicio
            })
            avancar(c)
            i += 1
            continue

        dicas = {
            "@": "Comece o comando com EMITIR, CHAMAR, INICIAR, FINALIZAR ou NAO_COMPARECEU.",
            "$": "Use P ou N seguido de três números para informar uma senha.",
            "!": "Para indicar ausência, use NAO_COMPARECEU seguido da senha."
        }
        erros.append({
            "caractere": c, "linha": linha_inicio, "coluna": coluna_inicio,
            "dica": dicas.get(c, "Remova esse caractere ou use um comando válido da fila.")
        })
        avancar(c)
        i += 1

    return tokens, erros

codigo_exemplo = """
EMITIR PREFERENCIAL
EMITIR NORMAL
CHAMAR P042 GUICHE G3
INICIAR P042
FINALIZAR P042
NAO_COMPARECEU N117 # senha não compareceu
@ CHAMAR $123
"""

tokens, erros = scanner_fila(codigo_exemplo)
print("TOKENS:")
for token in tokens:
    print(token)

print("\nERROS LÉXICOS:")
if not erros:
    print("Nenhum erro léxico.")
for erro in erros:
    print(f"Linha {erro['linha']}, coluna {erro['coluna']}: {erro['caractere']!r}. Dica: {erro['dica']}")
```

```python
# ==========================================
# 10. INTERFACE INTERATIVA DO PROJETO
# ==========================================

import ipywidgets as widgets
from IPython.display import display, clear_output

# Campos de entrada
campo_senha = widgets.Text(
    value="P042",
    description="Senha:",
    layout=widgets.Layout(width="380px")
)

campo_guiche = widgets.Text(
    value="G3",
    description="Guichê:",
    layout=widgets.Layout(width="380px")
)

# Botões
botao_senha = widgets.Button(
    description="Validar senha",
    button_style="info",
    layout=widgets.Layout(width="185px")
)

botao_guiche = widgets.Button(
    description="Validar guichê",
    button_style="info",
    layout=widgets.Layout(width="185px")
)

botao_scanner = widgets.Button(
    description="Executar scanner",
    button_style="success",
    layout=widgets.Layout(width="185px")
)

# Áreas de saída
saida_senha = widgets.Output()
saida_guiche = widgets.Output()
saida_scanner = widgets.Output()

# Caixa de comandos
campo_comandos = widgets.Textarea(
    value="EMITIR PREFERENCIAL\nCHAMAR P042 GUICHE G3",
    description="Comandos:",
    layout=widgets.Layout(width="100%", height="150px")
)


# ==========================================
# VALIDAR SENHA
# ==========================================

def ao_validar_senha(botao):
    with saida_senha:
        clear_output()

        if validar_senha(campo_senha.value):
            print("Senha válida.")
        else:
            print("Senha inválida.")


# ==========================================
# VALIDAR GUICHÊ
# ==========================================

def ao_validar_guiche(botao):
    with saida_guiche:
        clear_output()

        if validar_guiche(campo_guiche.value):
            print("Guichê válido.")
        else:
            print("Guichê inválido.")


# ==========================================
# SCANNER LÉXICO
# ==========================================

def executar_scanner(botao):
    with saida_scanner:
        clear_output()

        tokens = []
        erros = []

        # Palavras reservadas do projeto
        palavras_reservadas = {
            "EMITIR",
            "PREFERENCIAL",
            "NORMAL",
            "CHAMAR",
            "GUICHE"
        }

        # Analisa cada linha separadamente
        linhas = campo_comandos.value.splitlines()

        for numero_linha, linha in enumerate(linhas, start=1):
            posicao = 0

            while posicao < len(linha):

                # Ignora espaços
                if linha[posicao].isspace():
                    posicao += 1
                    continue

                # Lê a palavra ou o trecho até o próximo espaço
                inicio = posicao

                while (
                    posicao < len(linha)
                    and not linha[posicao].isspace()
                ):
                    posicao += 1

                lexema = linha[inicio:posicao]
                coluna = inicio + 1

                # Identifica palavras reservadas
                if lexema in palavras_reservadas:
                    tipo = "PALAVRA_RESERVADA"

                # Identifica senhas
                elif validar_senha(lexema):
                    tipo = "SENHA"

                # Identifica guichês
                elif validar_guiche(lexema):
                    tipo = "GUICHE"

                # Identifica outros identificadores
                elif re.fullmatch(
                    r"[A-Za-z_][A-Za-z_0-9]*",
                    lexema
                ):
                    tipo = "IDENTIFICADOR"

                # Identifica números
                elif re.fullmatch(r"[0-9]+", lexema):
                    tipo = "NUMERO"

                # Registra erros léxicos
                else:
                    erros.append({
                        "lexema": lexema,
                        "linha": numero_linha,
                        "coluna": coluna,
                        "dica": "Verifique o formato do comando."
                    })
                    continue

                tokens.append({
                    "tipo": tipo,
                    "lexema": lexema,
                    "linha": numero_linha,
                    "coluna": coluna
                })

        # Mostra os tokens
        print("TOKENS:")

        for token in tokens:
            print(token)

        # Mostra os erros
        print("\nERROS:")

        if erros:
            for erro in erros:
                print(erro)
        else:
            print("Nenhum erro léxico.")


# ==========================================
# CONECTAR OS BOTÕES ÀS FUNÇÕES
# ==========================================

botao_senha.on_click(ao_validar_senha)
botao_guiche.on_click(ao_validar_guiche)
botao_scanner.on_click(executar_scanner)


# ==========================================
# EXIBIR A INTERFACE
# ==========================================

display(campo_senha)
display(botao_senha)
display(saida_senha)

display(campo_guiche)
display(botao_guiche)
display(saida_guiche)

display(campo_comandos)
display(botao_scanner)
display(saida_scanner)


# ==========================================
# 11. TESTE DE MAXIMAL MUNCH NA INTERFACE
# ==========================================

def testar_maximal_munch(botao):
    with saida_scanner:
        clear_output()

        exemplos = [
            "EMITIR",
            "EMITIRX",
            "P042",
            "P0420",
            "G3",
            "G10",
            "NAO_COMPARECEU_EXTRA"
        ]

        print("TESTES DE MAXIMAL MUNCH")
        print("=" * 40)

        for exemplo in exemplos:
            print(f"\nEntrada: {exemplo}")
            print("Tokens:", scanner_maximal_munch(exemplo))


botao_maximal = widgets.Button(
    description="Testar Maximal Munch",
    button_style="warning",
    layout=widgets.Layout(width="230px")
)

botao_maximal.on_click(testar_maximal_munch)

display(botao_maximal)
```
